#### Aim: 
Calculate mass, SFR completeness limits for the VFS galaxies. Whereas Paper IV used MAGPHYS SED parameters, I will be using CIGALE parameters (see main notebook in the root of this directory). The procedures for evaluating these completeness limits is identical to that presented in Paper IV.

#### Purpose: 
Mstar and SFR completeness cuts limit the intrusion of selection bias in our sample. Stated differently, it ensures our sample of galaxies is representative of the true population and not biased by observational limitations. At different redshift slices, we do not want any one type of galaxy population to dominate, such as very bright/massive galaxies; rather, we want the slices to be as homogeneous as possible.

#### Note:
This section is NOT meant to be run as part of the feature clustering exercise, but rather to illustrate the process of finding the completeness limits for the VFS sample using CIGALE logSFR and logMstar columns.

#### Mass Completeness Steps (Pseudocode)

##### Summarized:
0. Apply all other 'cuts' to sample, if any. Mass completeness should come LAST.
    * sSFR, W3 S/N, size, morphology, etc. 

1. Select galaxies near z$_{max}$ = 0.025 (the upper end of our sample). Try to ensure ~400-500 galaxies in this collection.


2. From this group, select galaxies with -0.5 < m-m$_r$ < -1.0, where m$_r$ = 17.77 (SDSS apparent magnitude limit)
    * this bright subsample at the far end of our redshift distribution is far enough above the magnitude completeness limit so as to be equally complete for all r-band M/L. 


3. Compute $\Delta$m=m-m$_r$ and $\frac{L_{scale}}{L} = 10^{-0.4(-\Delta m)}$, where L$_{scale}$ is the scale luminosity. Note both luminosities are in the r-band. Then, calculate $\frac{M_{*,scale}}{M_*} = 10^{-0.4(-\Delta m)}$. 
    * we fade the galaxies and reduce their masses such that they lie at the m_r limit of our survey AND the ratios are unchanged: M/L = Mscale/Lscale
    * e.g., if the galaxy is 1 mag brighter than m_r, we reduce the stellar mass AND luminosity by a factor of 2.5. THIS is what those scaled variables do.
    * creates a synthetic pool of galaxies from this bright subsample that have the full range of M/L but are at our magnitude limit. 

4. L$_{scale}$, the luminosity of objects at apparent magnitude limit at the highest redshift limit of our sample, should be same for all galaxies in our subsample. DIFFERENT M$_{scale}$ --> we have a range of masses with ~same luminosity


5. Compute the distribution of M$_{*,scale}$ and find the upper 90th or 95th percentile --> corresponds to the ~highest M/L at $L_{scale}$
    * used the subsample to estimate how massive (M$_*$) a galaxy at our apparent magnitude limit must be in order to be detected regardless of M/L

-----

##### A Few Notes
* Without any corrections, we will only detect, close to our magnitude limit, low M/L galaxies (young stellar populations...much more luminosity for a given mass)
* THE GOAL is to find, at the upper redshift limit of our sample, the mass at and above which all M/L are visible
* Use bright subsample to estimate how massive a galaxy at our apparent magnitude limit must be in order to be detected regardless of M/L
  * To do so...calculate amount of fading required to bring each galaxy in the subsample to the magnitude limit of r=17.7
  * Then decrease the stellar masses by the same factor
  * The resulting stellar mass distribution has, by construction, identical apparent magnitudes and luminosities -- represents the intrinsic distribution in M/L (and hence in M*) for galaxies at our magnitude limit

#### Mass Completeness (for real)

##### 0. Load Tables

```python
cat = Table.read('data/cigale_vf_metallicity.fits')
vf = Table.read('data/virgowise_data.fits')
Z = Table.read('data/vf_v2_environment.fits')['Vcosmic']
r_mag = Table.read('data/vf_v2_r_photometry.fits')['mag_r']
Lr = Table.read('data/vf_v2_r_photometry.fits')['flux_r']

logmstar = np.log10(cat['bayes.stellar.m_star'])
logsfr = np.log10(cat['bayes.sfh.sfr'])
```

##### 1. Select a few galaxies near z$_{max}$. Try |z$_{max}$-z| < 0.4z$_{max}$.

```python
zflag = (Z>2500.) & (Z<3500)
```

```python
Z = Z[zflag]
r_mag = r_mag[zflag]
Lr = Lr[zflag]
logmstar = logmstar[zflag]
logsfr = logsfr[zflag]
```

2. From the group above, select galaxies with -0.5 < m-m$_r$ < 1.25
* m$_r$ = 17.77

```python
mr_flag = (r_mag.data > 16.52) & (r_mag.data < 17.27)
```

```python
Z = Z[mr_flag]
r_mag = r_mag[mr_flag]
Lr = Lr[mr_flag]
logmstar = logmstar[mr_flag]
logsfr = logsfr[mr_flag]
```

3. Compute the scale factors, M$_{scale}$ and L$_{scale}$

```python
L_scale = Lr.data * 10**(0.4*(r_mag.data-17.77))
```

```python
M_scale = (10**logmstar) * 10**(0.4*(r_mag.data-17.77))
```

4. Arrange distribution of M$_{scale}$, find upper 90-95th percentile

```python
#the ~M_scale.mask will remove any entries with no values
M_scale = np.sort(M_scale)
```

```python
#find 90% index 
index = int(len(M_scale)*0.9)
```

```python
#find M_scale corresponding to this index
M_limit = np.log10(M_scale[index])
```

```python
print(f'Mass Completeness Limit: {M_limit:.2f}')
```

#### Diagnostics

This section compares the MAGPHYS and CIGALE logSFR, logMstar model parameters for the VFS sample -- for my own visual satisfaction.

```python
os.chdir(homedir+'/github/machine-learning/galaxy_feature_clustering/')
cigale = Table.read('data/cigale_vf_metallicity.fits')
mstar_cig = np.log10(cigale['bayes.stellar.m_star'])
sfr_cig = np.log10(cigale['bayes.sfh.sfr'])

magphys = Table.read('data/vf-altphot.fits')
mstar_mag = magphys['combined_logMstar_med']
sfr_mag = magphys['combined_logSFR_med']

magphysflag = (magphys['magphys_flag'])&(sfr_mag!=0)
```

```python
sfr_cig = sfr_cig[magphysflag]
mstar_cig = mstar_cig[magphysflag]
mstar_mag = mstar_mag[magphysflag]
sfr_mag = sfr_mag[magphysflag]
```

```python
finite_mask = np.isfinite(sfr_cig) & np.isfinite(sfr_mag)
mean = np.nanmean(sfr_cig[finite_mask]-sfr_mag[finite_mask])

plt.scatter(sfr_mag, sfr_cig-sfr_mag ,s=10,alpha=0.4)
plt.axhline(0,color='black',alpha=0.5,label='1-to-1')
plt.axhline(mean,color='red',label=f'Mean Offset from 1-to-1\n{mean:.3f}')
plt.ylim(-5,5)
plt.legend()
plt.xlabel('[MAGPHYS logSFR]',fontsize=12)
plt.ylabel('[CIGALE logSFR] - [MAGPHYS logSFR]',fontsize=12)
```

```python
plt.scatter(mstar_cig, sfr_cig, s=10,alpha=0.4,color='black',label='CIGALE')
plt.scatter(mstar_mag, sfr_mag, s=10,alpha=0.1,color='pink',label='MAGPHYS')
plt.xlim(5,)
plt.ylim(-10,)
plt.xlabel('logMstar',fontsize=12)
plt.ylabel('logSFR',fontsize=12)
plt.legend()
```
